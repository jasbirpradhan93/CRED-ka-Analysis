## Setup

## SWOT Data

```
swot = {
    'Strengths': [
        'Unique Value Proposition (Gamified credit card bill rewards)',
        'Strong Brand Identity & Premium Positioning',
        'Loyal Affluent User Base with High Creditworthiness',
        'Excellent Customer Experience (UI/UX, No Ads)',
        'Robust Tech & Data Infrastructure'
    ],
    'Weaknesses': [
        'Limited User Base (~35M credit card users vs 300M+ UPI)',
        'High Dependence on Rewards to Drive Engagement',
        'Profitability Challenges (High losses despite revenue growth)'
    ],
    'Opportunities': [
        'Expand into Loans, BNPL, Insurance, Wealth products',
        'Data-driven personalization with AI/ML',
        'Partnerships with e-commerce, lifestyle, travel',
        'International Expansion into SE Asia, NRI markets'
    ],
    'Threats': [
        'Intense Competition in Indian fintech',
        'Regulatory Risks (RBI, SEBI tightening)',
        'User Fatigue from gamified rewards',
        'Credit Card Market Shifts (BNPL rise)',
        'Economic Downturns affecting discretionary spending'
    ]
}
pd.DataFrame(dict([(k,pd.Series(v)) for k,v in swot.items()]))
```

## Visual: SWOT Quadrant

```
plot_swot_quadrant(swot['Strengths'], swot['Weaknesses'], swot['Opportunities'], swot['Threats'], save_path='swot_quadrant.png')
```

## Insights
- **Strengths**: Premium positioning, loyal affluent base, and strong brand campaigns like Rahul Dravid’s ad.
- **Weaknesses**: Niche user base and unsustainable reward-heavy engagement.
- **Opportunities**: Expand into broader financial services and leverage data for personalization.
- **Threats**: Heavy competition, BNPL disruption, and regulatory risks.
